# Datagen EDA (Phase 4 step 7)

Sanity plots over the synthetic dataset: seasonal decomposition, occupancy histogram,
price–demand scatter, snapshot correlation heatmap, and event-uplift views.

**Prerequisite:** generate data first, e.g.

```bash
uv run python -m datagen run --config datagen/configs/tiny.yaml --output ../../artifacts/datasets/tiny
```

Or headless figures: `uv run python -m datagen eda --input … --figures …`.


````python
from pathlib import Path

from IPython.display import Image, display

from datagen.config import DatagenConfig
from datagen.eda import run_eda

REPO = Path("../..").resolve()
SERVICE = Path("..").resolve()

# Prefer default dataset; fall back to tiny.
DATASET = REPO / "artifacts" / "datasets" / "default"
CONFIG = SERVICE / "datagen" / "configs" / "default.yaml"
FIGURES = REPO / "artifacts" / "eda" / "notebook"

if not (DATASET / "nights.parquet").is_file():
    DATASET = REPO / "artifacts" / "datasets" / "tiny"
    CONFIG = SERVICE / "datagen" / "configs" / "tiny.yaml"
    FIGURES = REPO / "artifacts" / "eda" / "tiny"

config = DatagenConfig.from_yaml(CONFIG)
rooms = {rt.code: rt.rooms for rt in config.hotel.room_types}
result = run_eda(dataset_dir=DATASET, figures_dir=FIGURES, rooms_by_type=rooms)
print(f"dataset={DATASET}")
print(f"occupancy_mean={result.occupancy_mean:.2%}")
for name, path in result.figure_paths.items():
    print(name, "->", path)
    display(Image(filename=str(path)))
````
